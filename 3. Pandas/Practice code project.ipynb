```python
import numpy as np
import pandas as pd

# Table 1: Sensor readings from 8 robotic trials
trials = pd.DataFrame({
    'trial_id': ['T1', 'T2', 'T3', 'T4', 'T5', 'T6', 'T7', 'T8'],
    'robot_type': ['arm', 'drone', 'arm', 'rover', 'drone', 'arm', 'rover', 'drone'],
    'battery_pct': ['92%', '85%', np.nan, '78%', '90%', '88%', '81%', np.nan],
    'reward': list((14.5, 22.0, 11.0, 18.5, 25.0, 15.5, 19.0, 21.5)),
    'status': ['ok', 'ok', 'fail', 'ok', 'ok', 'ok', 'fail', 'ok']
})

# Table 2: Hardware weight specs per robot_type
specs = pd.DataFrame({
    'robot_type': ['arm', 'drone', 'rover'],
    'mass_kg': list((12.5, 2.4, 18.0))
})

def func(num):
  if pd.isna(num):
    return np.nan
  num=str(num)
  num=num[0:2]
  num=float(num)
  return num
def find(rob):
  return specs[rob]#or .loc
trials=trials[trials['status']=='ok']#Broadcasting
trials.battery_pct=trials.battery_pct.apply(func)# mapping
trials.battery_pct=trials.battery_pct.fillna(trials.battery_pct.mean())#filling all nans

specs=specs.set_index('robot_type')#converting index of specs
specs=specs.squeeze()#squeeze--converting into series
trials['mass_kg']=trials['robot_type'].apply(find)#we have to map all robot type using spec as a function to the corresponding mass kg of each type
trials=trials.set_index("trial_id")
#combined = df1.join(df2, on='matching_col', how='left')
#result = df.join(other, on=None, how='left', lsuffix='', rsuffix='')

trials['efficiency']=trials.reward/trials.mass_kg
grouped=trials.groupby('robot_type')['efficiency'].agg(['mean'])
grouped2=trials.groupby('robot_type')['reward'].agg(['max'])
summary=grouped.join(grouped2)
summary=pd.concat([grouped,grouped2],axis=1)#NEED to specify the axis
#Join-ONLY side by side, Concat: depend on axis
# summary = trials.groupby('robot_type').agg({'efficiency': 'mean', 'reward': 'max'}) mapping efficiency to mean, reward to max.

summary=summary.sort_values(["mean","max"],ascending=False)
#Sorting first by mean, then by max, in descending values
y = trials['reward'].to_numpy(dtype=np.float64)
#Converting to array Same for y=array(trials['reward'])
features_df = trials.drop(columns=['reward','status'])
#Dropping 2 columns
features_df = pd.get_dummies(features_df, columns=['robot_type'], dtype=float)
#converting types into a dummy
features_df = np.array(features_df)
#
summary
```
