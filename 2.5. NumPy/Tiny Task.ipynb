```python
"""
The Challenge: 2D Thermal Grid & Hotspot Analyzer
Imagine a 10 x 10 grid of temperature sensors recording heat readings over 24 hours.
Some sensors glitch and output invalid readings (-999.0).
Your job is to clean the data and detect sudden heat spikes using pure NumPy (no for loops)."""
import numpy as np
def get_ave(arr):
    total = (
        arr[:, 0:8, 0:8] + arr[:, 0:8, 1:9] + arr[:, 0:8, 2:10] +
        arr[:, 1:9, 0:8] + arr[:, 1:9, 1:9] + arr[:, 1:9, 2:10] +
        arr[:, 2:10, 0:8] + arr[:, 2:10, 1:9] + arr[:, 2:10, 2:10]
    )
    return total / 9.0
np.random.seed(7)

# Shape: (24 hours, 10 rows, 10 cols) — baseline temps around 20°C to 35°C
temps = np.random.uniform(20.0, 35.0, size=(24, 10, 10))

# Inject 15 random sensor glitches (-999.0)
glitch_idx = tuple(np.random.randint(0, s, size=15) for s in temps.shape)
temps[glitch_idx] = -999.0

# Inject a 3x3 heatwave spike at hour 14 in the center of the grid
temps[14, 4:7, 4:7] += 18.0

test=temps.copy()
temps[temps<=-998.0]=np.nan
sensor_means = np.nanmean(temps, axis=0, keepdims=True)
temps = np.where(np.isnan(temps), sensor_means, temps)
time_means = np.sum(temps,(1,2))/100
temps = temps-time_means.reshape(24,1,1)
spike=np.argmax(temps)
specific=np.unravel_index(spike,(24,10,10))
print(specific)
average=np.zeros((24,8,8))
average=get_ave(temps)
```
